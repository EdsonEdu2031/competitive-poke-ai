```python
# Código para adicionar a Meteor que não deu pra incluir no Cluster
# Irá abrir arquivo por arquivo, adicionar a métrica e salvar

import nltk
import pandas as pd
from nltk.translate.meteor_score import meteor_score

models = ['DeepSeek', 'Gemma', 'Llama', 'Qwen']

for modelo in models:
    df = pd.read_csv(f'C:/PokeAI/databases/P&R/Metagame/2-Automatic/Sem_Meteor/Eval_Automatic_{modelo}_Perguntas_Metagame.csv')
    for indice, row in df.iterrows():
        referencia = row['resposta']
        predicao = row['resposta_gerada']
        nota_meteor = meteor_score([referencia.split()], predicao.split())
        df.at[indice, 'meteor'] = nota_meteor
    df.to_csv(f'C:/PokeAI/databases/P&R/Metagame/2-Automatic/Eval_Automatic_{modelo}_Perguntas_Metagame.csv', index=False)
```
