```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv('C:/PokeAI/databases/Perguntas_Metagame.csv')

df_train, df_test = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    shuffle=True
)

# Reiniciar os índices
df_train = df_train.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

df_train.to_csv('C:/PokeAI/databases/Perguntas_Metagame_train.csv', index=False)
df_test.to_csv('C:/PokeAI/databases/Perguntas_Metagame_test.csv', index=False)

print(f"Total: {len(df)}")
print(f"Treino: {len(df_train)} ({len(df_train) / len(df) * 100:.1f}%)")
print(f"Teste: {len(df_test)} ({len(df_test) / len(df) * 100:.1f}%)")
```
