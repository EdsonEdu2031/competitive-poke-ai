```python
import pandas as pd
```

```python
pokemons_df = pd.read_json("../databases/smogon_metagame_data/chaos_gen9championsvgc2026regmbbo3-1760.json")
dados_por_pokemon = pokemons_df["data"].dropna()

info = ["Abilities", "Items", "Spreads", "Moves", "Teammates"]
pokemon_nome = "Emolga"

for i in info:
    print(i, dados_por_pokemon[pokemon_nome][i])
```

```python
teammates = dados_por_pokemon[pokemon_nome]["Teammates"]
FATOR_ESCALA = 25.1675

teammates_porcentagem = {
    poke: valor * FATOR_ESCALA for poke, valor in teammates.items()
}

teammates_ordenados = sorted(
    teammates_porcentagem.items(), key=lambda x: x[1], reverse=True
)

print(f"Correlação entre {pokemon_nome} e teammates pelo total: ")
for poke, porcentagem in teammates_ordenados:
  if porcentagem > 0.05:
    print(f"{poke}: {porcentagem:.2f}%")
```

```python
dados_especificos = dados_por_pokemon[pokemon_nome]

peso_total_pokemon = sum(dados_especificos["Abilities"].values())

teammates_series = pd.Series(dados_especificos["Teammates"])

teammates_porcentagem = (teammates_series / peso_total_pokemon) * 100

teammates_ordenados = teammates_porcentagem.sort_values(ascending=False)

print(f"Correlação entre {pokemon_nome} e teammates pelo total: ")
for poke, porcentagem in teammates_ordenados.items():
    if porcentagem > 0.05:
        print(f"{poke}: {porcentagem:.2f}%")
```
