```python
import pandas as pd
```

```python
pokemons_df = pd.read_json("../databases/smogon_metagame_data/chaos_gen9championsvgc2026regmbbo3-1760.json")
dados_por_pokemon = pokemons_df["data"].dropna()

info = ["Abilities", "Items", "Spreads", "Moves", "Teammates"]
pokemon_nome = "Emolga"

for i in info:
    print(i, dados_por_pokemon[pokemon_nome][i])
```

```python
teammates_dict = dados_por_pokemon[pokemon_nome]["Teammates"]

total_aparicoes = sum(teammates_dict.values())

teammates_porcentagem = {poke: (valor / total_aparicoes) * 100 for poke, valor in teammates_dict.items()}

teammates_ordenados = sorted(
    teammates_porcentagem.items(), key=lambda x: x[1], reverse=True
)

print(f"Correlação entre {pokemon_nome} e teammates pelo total: ")
for poke, porcentagem in teammates_ordenados:
    if porcentagem > 0.05:
        print(f"{poke}: {porcentagem:.2f}%")
```

```python
raw_count_principal = dados_por_pokemon[pokemon_nome]["Raw count"]
teammates_dict = dados_por_pokemon[pokemon_nome]["Teammates"]

teammates_uso_proprio = {poke: (valor / raw_count_principal) * 100 for poke, valor in teammates_dict.items()}

teammates_uso_ordenado = sorted(teammates_uso_proprio.items(), key=lambda x: x[1], reverse=True)

print(f"Correlação entre {pokemon_nome} e teammates em relação ao pokemon ({raw_count_principal}): ")
for poke, porcentagem in teammates_uso_ordenado:
  print(f"{poke}: {porcentagem:.2f}%")
```
